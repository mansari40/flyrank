## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

df = df[
    (df["impressions_90d"] > 0) &
    (df["content_age_days"] >= 90)
].drop_duplicates("content_id")

print(f"Usable pages: {len(df):,}")
print(f"Pages with a downward trend: {(df['trend_direction'] == 'down').mean():.1%}")
print(f"Median 90-day impressions: {df['impressions_90d'].median():,.0f}")
display(df.head())
```

The starter data gives me a good reason to look further into this problem. After applying the starter filters, there are 30,000 usable pages. 54.2% of these pages have a downward trend, based on the current trend label. The median page has 731 impressions over 90 days, so there is enough search activity to compare pages and identify which ones may deserve attention first. These numbers do not prove that a page needs a refresh, but they make the prioritization problem worth investigating.
